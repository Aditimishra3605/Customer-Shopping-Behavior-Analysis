## Changing col names into snake case

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

## Feature Engineering

```python
# creating column age_grp

labels = ['Young','Adult','Middle-aged','Senior']
df['age_grp']=pd.qcut(df['age'],q=4,labels=labels)

'''This code snippet divides a continuous numerical column(age) into four equally sized groups (quartiles) 
based on its values and assigns custom labels to each group. 
The qcut function stands for Quantile-based discretization. It looks at all the ages in the dataset, sorts them 
from youngest to oldest, and calculates the cut-off points so that each of the 4 groups contains 
roughly the same number of data points (25% of the data per group).'''
```

```python
df['frequency_of_purchases'].unique()
```

```python
# creating column purchase_freq_days
freq_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_freq_days']=df['frequency_of_purchases'].map(freq_mapping)
```

if the 'review ratings' column contained null values:
we would have filled the null values of each category with median of that particular category. 
reason: mean can get affected by outliers and that would skew the distribution of our data. But, median is robust to outliers.

df['Review Rating']=df.groupby('Category')['Review rating'].transform(lambda x: x.fillna(x.median()))
